```python
import numpy as np

# 1. Наша целевая функция (яма)
def F(x1, x2):
    eq1 = x2 - np.sin(x1)
    eq2 = x1**2 + x2**2 - 1
    return eq1**2 + eq2**2

def scan_minimize(func, x_start, epsilon_scan):
    x = x_start
    h = 0.1

    while abs(h) > epsilon_scan:
        if func(x + h) >= func(x):
            if func(x - h) < func(x):
                h = -h
            else:
                h = h / 10
                continue

        x_next = x + h
        while func(x_next + h) < func(x_next):
            x_next += h

        x = x_next
        h = -h / 10

    return x

epsilon = 1e-2
x1, x2 = 0.7, 0.7
max_iter = 100
iteration = 0

print(f"Старт: x1 = {x1}, x2 = {x2}")

while iteration < max_iter:
    x1_old, x2_old = x1, x2

    x1 = scan_minimize(lambda x: F(x, x2), x1_old, epsilon)
    x2 = scan_minimize(lambda y: F(x1, y), x2_old, epsilon)
    iteration += 1

    step_size = np.sqrt((x1 - x1_old)**2 + (x2 - x2_old)**2)
    if step_size < epsilon:
        print(f"\nАлгоритм сошелся за {iteration} итераций!")
        break

print("-" * 30)
print(f"Ответ: x1 = {x1:.4f}, x2 = {x2:.4f}")
print(f"Значение F(x1, x2) (погрешность): {F(x1, x2):.6f}")

if x1 > 0:
    print("Условие x1 > 0 выполнено.")
else:
    print("Условие x1 > 0 не выполнено.")
```
