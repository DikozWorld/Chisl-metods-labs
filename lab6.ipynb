```python
import numpy as np

x1 = 0.7
x2 = 0.7
epsilon = 1e-2
max_iter = 100
iteration = 0

print(f"Старт: x1 = {x1}, x2 = {x2}")

while iteration < max_iter:
    x1_old, x2_old = x1, x2

    x1_new = np.sqrt(1 - x2_old**2)
    x2_new = np.sin(x1_old)

    x1, x2 = x1_new, x2_new
    iteration += 1

    step_size = np.sqrt((x1 - x1_old)**2 + (x2 - x2_old)**2)

    if step_size < epsilon:
        print(f"\nАлгоритм сошелся за {iteration} итераций!")
        break

print("-" * 30)
print(f"Ответ: x1 = {x1:.6f}, x2 = {x2:.6f}")
```
